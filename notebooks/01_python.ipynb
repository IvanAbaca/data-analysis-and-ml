```
Usá zip() y un for para recorrer ambas listas simultáneamente e imprimir:
```

Ejercicio 9 — Bloques e indentación

Partí de:
```python
score = 85
submitted = True
```
Hacé un programa que:

Si submitted es True, evalúe el score:
- 90 o más → "Excellent"
- 60 o más → "Approved"
- menos de 60 → "Failed"
- Si submitted es False, imprima "Not submitted".

````python
score = 85
submitted = True

def evaluate_score(score: int, submitted: bool) -> None:
    if not submitted:
        print("Not submitted")
        return

    if score >= 90:
        print("Excellent")
    elif score >= 60:
        print("Approved")
    else:
        print("Failed")

evaluate_score(score, submitted)
````

Ejercicio 12 — Funciones

Creá una función llamada:
```python
def calculate_average()
```
que:

- reciba una lista de números;
- calcule su promedio;
- devuelva el resultado usando return;
- fuera de la función, guardá el resultado en una variable llamada average;
- imprimí average.

Probala con:
```python
scores = [80, 65, 90, 75, 100]
```

````python
def calculate_average(numbers: list[int | float]) -> float:
    total = 0

    for number in numbers:
        total += number

    return total / len(numbers)

scores = [80, 65, 90, 75, 100]
average = calculate_average(scores)

print(average)
````

Ejercicio 14 — lambda como criterio de ordenamiento

Partí de:
```python
students = [
    ("Ana", 75),
    ("Ivan", 90),
    ("Lucas", 60),
    ("Sofia", 85),
]
```
Ordená la lista de menor a mayor según la nota, usando:
```python
sort(key=...)
```
y pasando una lambda como criterio.

Después imprimí students.

Resultado esperado:
```python
[('Lucas', 60), ('Ana', 75), ('Sofia', 85), ('Ivan', 90)]
```

````python
students = [
    ("Ana", 75),
    ("Ivan", 90),
    ("Lucas", 60),
    ("Sofia", 85),
]

print(students.sort(key=lambda x: x[0]))
````
