```python
import pandas as pd
bike_buyers = pd.read_csv('https://raw.githubusercontent.com/pokengineer/DataScience/main/datasets/bike_buyers.csv')
```

1. ¿Cuales de las variables contienen valores nulos?

```python
bike_buyers[bike_buyers.isna().any(axis=1)]
```

2. ¿Está Balanceada la variable target 'Purchased Bike'?

```python
bike_buyers['Purchased Bike'].value_counts().to_frame()
```

3. Imputar los valores nulos de Gender con el valor 'X'.

```python
bike_buyers['Gender'] = bike_buyers['Gender'].fillna('X')
```

4. Imputar los valores nulos y outliers de Age con la media.

```python
age = bike_buyers['Age']

q1, q3 = age.quantile([0.25, 0.75])
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

mean_age = age.mean()

invalid_age = age.isna() | ~age.between(lower_bound, upper_bound)

bike_buyers['Age'] = age.mask(invalid_age, mean_age)
```

5. Eliminar los registros con valor nulo en Cars.

```python
# bike_buyers_without_car = bike_buyers[bike_buyers['Cars'].notna()]
bike_buyers_without_car = bike_buyers.dropna(subset=['Cars'])
```

6. Generar columnas dummy para las variables Education y Occupation.

```python
education_dummies = pd.get_dummies(bike_buyers['Education'])
occupation_dummies = pd.get_dummies(bike_buyers['Occupation'])
```
