## Ejercicio 7 — `iloc`: selección por posición

Usá `iloc` para seleccionar una fila por posición y un valor indicando fila y columna por posición.


````python
import pandas as pd

students = pd.DataFrame([
    {"name": "Ana", "age": 20, "score": 70},
    {"name": "Ivan", "age": 30, "score": 75},
    {"name": "Lucas", "age": 40, "score": 72},
])

print(students.iloc[1, 0])
````

## Ejercicio 8 — `loc`: selección por etiqueta

Usá índices personalizados y `loc` para seleccionar una fila por etiqueta y un valor por etiqueta de fila y nombre de columna.


````python
import pandas as pd

students = pd.DataFrame([
        {"name": "Ana", "age": 20, "score": 70},
        {"name": "Ivan", "age": 30, "score": 75},
        {"name": "Lucas", "age": 40, "score": 72},
    ], index=["student_a", "student_b", "student_c"]
)

print(students.loc["student_a"])
print(students.loc["student_b", "name"])
````

## Ejercicio 9 — Slicing

Usá `iloc` para obtener un subconjunto de filas y columnas.


````python
import pandas as pd

students = pd.DataFrame([
        {"name": "Ana", "age": 20, "score": 70},
        {"name": "Ivan", "age": 30, "score": 75},
        {"name": "Lucas", "age": 40, "score": 72},
    ], index=["student_a", "student_b", "student_c"]
)

print(students.iloc[:2, 2:])
````

## Ejercicio 10 — Filtros

Conservá únicamente las filas que cumplan una condición numérica.


````python
import pandas as pd

students = pd.DataFrame([
        {"name": "Ana", "age": 20, "score": 70},
        {"name": "Ivan", "age": 30, "score": 75},
        {"name": "Lucas", "age": 40, "score": 72},
    ], index=["student_a", "student_b", "student_c"]
)

print(students[students["age"] > 29])
````

## Ejercicio 11 — Múltiples condiciones

Filtrá combinando dos condiciones sobre las columnas.


````python
import pandas as pd

students = pd.DataFrame([
        {"name": "Ana", "age": 20, "score": 70},
        {"name": "Ivan", "age": 30, "score": 75},
        {"name": "Lucas", "age": 40, "score": 72},
    ], index=["student_a", "student_b", "student_c"]
)

print(students[(20 < students["age"]) & (students["age"] < 40)])
````

## Ejercicio 12 — Agregaciones

Calculá suma, media, mínimo y máximo sobre una columna numérica.


````python
import pandas as pd

students = pd.DataFrame([
        {"name": "Ana", "age": 20, "score": 70},
        {"name": "Ivan", "age": 30, "score": 75},
        {"name": "Lucas", "age": 40, "score": 72},
    ], index=["student_a", "student_b", "student_c"]
)

print(students["age"].sum())
print(students["age"].mean())
print(students["age"].min())
print(students["age"].max())
````
